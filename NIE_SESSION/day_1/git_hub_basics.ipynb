
# 🚀 Introduction to GitHub

**GitHub** is a cloud-based platform that helps developers **store, manage, and collaborate on code** using **Git**, a version control system.

It allows you to:

* Track changes in your code over time
* Work with teams without overwriting each other’s work
* Maintain multiple versions (branches) of a project
* Host and share projects publicly or privately

In simple terms:
👉 **Git = version control tool**
👉 **GitHub = platform to store and collaborate on Git repositories**

---

# 🔑 Key Concepts

* **Repository (Repo)** → Project folder with code
* **Commit** → Snapshot of changes
* **Branch** → Separate version of code
* **Merge** → Combine changes
* **Pull Request (PR)** → Request to merge changes
* **Clone** → Copy repo to local system
* **Push** → Upload code to GitHub
* **Pull** → Get latest code from GitHub

---

# ⚡ Important Git Commands

## 🔹 Setup

```bash
git config --global user.name "Your Name"
git config --global user.email "your@email.com"
```

---

## 🔹 Initialize Repo

```bash
git init
```



## 🔹 Clone Repository

```bash
git clone https://github.com/username/repo.git


## 🔹 Check Status

```bash
git status
```



## 🔹 Add Files

```bash
git add file.txt
git add .
```


## 🔹 Commit Changes

```bash
git commit -m "your message"
```


## 🔹 Push to GitHub

```bash
git push origin main
```


## 🔹 Pull Latest Changes

```bash
git pull origin main
```


## 🔹 Branching

```bash
git branch new-branch
git checkout new-branch
```

OR (shortcut)

```bash
git checkout -b new-branch
```


## 🔹 Merge Branch

```bash
git checkout main
git merge new-branch
```


## 🔹 View Commit History

```bash
git log
```
